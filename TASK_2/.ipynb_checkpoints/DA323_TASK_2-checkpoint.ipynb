# Multimodal Matching: Video & Audio Correlation Analysis

**Overview:**  
In this notebook, we process a dataset containing 45 simulated videos (showing a ball in projectile motion) and 45 audio tracks.  
- **Video Processing:** We extract the ball's position frame-by-frame using OpenCV, then compute the vertical (y-axis) velocity. A change in the sign of the y velocity is used as an indicator of a collision (e.g., when the ball bounces off a boundary).  
- **Audio Processing:** We load each audio file with librosa and use an onset detection algorithm to extract event timings that should correspond to the ball collisions.  
- **Matching:** We compute a DTW (Dynamic Time Warping) distance between the collision times (from video) and onset times (from audio). These distances are assembled into a cost matrix, and the Hungarian algorithm (linear sum assignment) is used to find the optimal one-to-one match.

The final matches are then saved to a CSV file.

## Step 0: Import Libraries

We first import the required libraries for video processing, audio processing, numerical computations, and plotting.


```python
import cv2
import numpy as np
import os
import glob
import librosa
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import linear_sum_assignment
```

## Step 1: Video Processing – Extracting Ball Positions and Collision Times

The following function `process_video` takes a video file path as input and:

1. Opens the video with OpenCV.
2. Processes each frame to detect the ball using Hough Circle detection (assuming the ball appears as a circular object).
3. Records the ball's (x, y) position and the corresponding timestamp.
4. Computes the difference in the y-coordinate between consecutive frames (i.e., vertical velocity).
5. Detects collisions when the sign of the vertical velocity changes.

*Note:* Depending on your simulated videos, you might need to adjust the Hough Circle parameters (like `minRadius` and `maxRadius`) for optimal ball detection.


### Step 1.1: Defining the process_video function

```python
def process_video(video_path):
    """
    Processes a video file to extract the ball's collision times.
    
    Args:
        video_path (str): Path to the video file.
        
    Returns:
        collision_times (list of float): Timestamps where a collision is detected.
        all_timestamps (list of float): Timestamps for all frames where ball position is detected.
        positions (list of tuples): Detected (x, y) positions of the ball.
    """
    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    positions = []
    timestamps = []
    frame_idx = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        # Convert frame to grayscale
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        # Apply Gaussian blur to reduce noise
        blurred = cv2.GaussianBlur(gray, (7, 7), 0)
        # Detect circles using HoughCircles
        circles = cv2.HoughCircles(blurred, cv2.HOUGH_GRADIENT, dp=1.2, minDist=50,
                                   param1=50, param2=30, minRadius=5, maxRadius=30)
        if circles is not None:
            circles = np.round(circles[0, :]).astype("int")
            # Assume the first detected circle corresponds to the ball
            ball = circles[0]  # (x, y, r)
            positions.append((ball[0], ball[1]))
        else:
            # If no ball detected, append None
            positions.append(None)
        timestamps.append(frame_idx / fps)
        frame_idx += 1
        
    cap.release()
    
    # Filter out frames where ball detection failed
    valid_data = [(t, pos) for t, pos in zip(timestamps, positions) if pos is not None]
    if not valid_data:
        return [], [], []
    times, pos_list = zip(*valid_data)
    
    # Compute y-velocity from the y positions
    y_positions = [pos[1] for pos in pos_list]
    y_velocity = np.diff(y_positions)
    collision_times = []
    
    # Detect collisions by checking sign changes in y-velocity
    for i in range(1, len(y_velocity)):
        if y_velocity[i-1] * y_velocity[i] < 0:
            # Approximate collision time using the current timestamp
            collision_times.append(times[i])
    
    return collision_times, times, pos_list
```


## Step 2: Audio Processing – Extracting Onset Times

The following function `process_audio` loads an audio file and uses librosa's onset detection to extract times when significant audio events occur.

*Note:* You can experiment with onset detection parameters (e.g., `hop_length` or `backtrack`) to improve detection accuracy.


### Step 2.1: Defining the process_audio function

```python
def process_audio(audio_path):
    """
    Processes an audio file to extract onset times.
    
    Args:
        audio_path (str): Path to the audio file.
        
    Returns:
        onset_times (ndarray): Array of detected onset times (in seconds).
    """
    y, sr = librosa.load(audio_path, sr=None)
    # Detect onset times using librosa
    onset_times = librosa.onset.onset_detect(y=y, sr=sr, units='time')
    return onset_times
```

### Step 2.2: **Test the audio processing on a sample audio file.**  
Ensure that your `audio_only` folder contains the audio files.

```python
audio_files = sorted(glob.glob(r"C:\Users\iitia\Downloads\DA323_project\TASK_2\dataset\audio_only/*.wav"))
if audio_files:
    sample_audio = audio_files[0]
    onset_times = process_audio(sample_audio)
    print("Sample Audio:", os.path.basename(sample_audio))
    print("Detected Onset Times:", onset_times)
else:
    print("No audio files found in 'audio_only' directory.")
```

## Step 3: Feature Comparison and Matching

-- To match each video with its corresponding audio track, we compare the sequence of collision times (from video) with the sequence of onset times (from audio).  

- We define a simple DTW (Dynamic Time Warping) function to compute a distance between the two sequences.  
- Then, we build a cost matrix where each element \(\text{cost}_{ij}\) is the DTW distance between the collision times of the \  (i^{th}\) video and the onset times of the \(j^{th}\) audio file.

- Finally, we solve the assignment problem using the Hungarian algorithm to get the optimal one-to-one mapping.


### Step 3.1: Defining the compute_dtw_distance fucntion

```python
def compute_dtw_distance(video_events, audio_events):
    """
    Computes a DTW-based distance between two sequences of event times.
    
    Args:
        video_events (list of float): Collision times from video.
        audio_events (list or array of float): Onset times from audio.
    
    Returns:
        dtw_distance (float): The DTW distance between the two sequences.
    """
    if len(video_events) == 0 or len(audio_events) == 0:
        return np.inf
    v = np.array(video_events).reshape(-1, 1)
    a = np.array(audio_events).reshape(-1, 1)
    
    # Create a cost matrix based on absolute differences
    n, m = len(v), len(a)
    cost = np.zeros((n, m))
    for i in range(n):
        for j in range(m):
            cost[i, j] = abs(v[i] - a[j])
    
    # Initialize DTW matrix
    dtw = np.zeros((n, m))
    dtw[0, 0] = cost[0, 0]
    # Initialize first row and column
    for i in range(1, n):
        dtw[i, 0] = cost[i, 0] + dtw[i-1, 0]
    for j in range(1, m):
        dtw[0, j] = cost[0, j] + dtw[0, j-1]
        
    # Populate DTW matrix
    for i in range(1, n):
        for j in range(1, m):
            dtw[i, j] = cost[i, j] + min(dtw[i-1, j], dtw[i, j-1], dtw[i-1, j-1])
            
    return dtw[-1, -1]
```

### Step 3.2 :
- Next, we build the cost matrix for all pairs of video and audio files.  
- For each video file, we extract the collision times, and for each audio file, we extract the onset times.  
- Then we compute the DTW distance between every pair.

```python
def build_cost_matrix(video_files, audio_files):
    n_videos = len(video_files)
    n_audios = len(audio_files)
    cost_matrix = np.zeros((n_videos, n_audios))
    
    # Precompute features for each video and audio file
    video_features = []
    for vfile in video_files:
        collisions, _, _ = process_video(vfile)
        video_features.append(collisions)
        
    audio_features = []
    for afile in audio_files:
        onset_times = process_audio(afile)
        audio_features.append(onset_times)
    
    # Compute DTW distances
    for i in range(n_videos):
        for j in range(n_audios):
            cost_matrix[i, j] = compute_dtw_distance(video_features[i], audio_features[j])
    
    return cost_matrix
```

**Build the cost matrix:**

```python
# Build the cost matrix
if video_files and audio_files:
    cost_matrix = build_cost_matrix(video_files, audio_files)
    print("Cost Matrix Shape:", cost_matrix.shape)
else:
    print("Make sure both video and audio files are available.")
```

**Now, we solve the assignment problem using the Hungarian algorithm to determine the optimal matching between video and audio files.**


```python
# Solve the assignment problem (Hungarian algorithm)
row_ind, col_ind = linear_sum_assignment(cost_matrix)
```

## Step 4: Saving Final Matches

- We create a CSV file (`final_matches.csv`) with two columns: `video_file` and `audio_file`.
- Each row represents a match determined by the assignment algorithm.

```python
matches = []
for i, j in zip(row_ind, col_ind):
    video_name = os.path.basename(video_files[i])
    audio_name = os.path.basename(audio_files[j])
    matches.append((video_name, audio_name))

matches_df = pd.DataFrame(matches, columns=["video_file", "audio_file"])
matches_df.to_csv("final_matches.csv", index=False)
print("Final matches saved to 'final_matches.csv'.")
print(matches_df)
```

## Step 5: Visualizing and Analyzing the Results

- You might want to visualize the collision times and onset times for a given video–audio pair to inspect the alignment.

- The following cell plots the detected events for the first matched pair.

```python
if video_files and audio_files:
    # Use the first matched pair for visualization
    idx_video = row_ind[0]
    idx_audio = col_ind[0]
    
    collisions, t_video, _ = process_video(video_files[idx_video])
    onset_times = process_audio(audio_files[idx_audio])
    
    plt.figure(figsize=(10, 4))
    plt.plot(t_video, [0]*len(t_video), 'ko', markersize=2, label='Frame Times')
    plt.vlines(collisions, -0.5, 0.5, color='r', linewidth=2, label='Collision Times (Video)')
    plt.vlines(onset_times, -1, 1, color='b', linestyles='dashed', label='Onset Times (Audio)')
    plt.xlabel('Time (s)')
    plt.legend()
    plt.title('Event Timing Comparison for a Matched Pair')
    plt.show()
else:
    print("No files available for visualization.")
```

## Conclusion

In this notebook, we implemented a multimodal matching pipeline:

- **Video Analysis:** Extracted ball positions and detected collision events based on changes in the vertical velocity.
- **Audio Analysis:** Extracted onset events using librosa.
- **Matching:** Computed DTW distances between event sequences and solved the matching problem with the Hungarian algorithm.

The final output is a CSV file (`final_matches.csv`) containing the matched pairs of video and audio files. This approach can be refined further by improving the detection algorithms or using more sophisticated feature representations.

Feel free to experiment with different parameters and methods to further improve the matching accuracy.


```python
# %% [markdown]
# ## Additional Cell: Combine and Display Video with Audio in Notebook
#
# This cell demonstrates how to merge a muted video from the `video_only` folder with its corresponding audio track from the `audio_only` folder.
# We use MoviePy to load the video and audio clips, assign the audio to the video, save the combined clip to a temporary file, and finally display it inline using IPython's Video display.
#
# **Note:** Merging and writing the video may take a few seconds depending on your system.

# %%
from moviepy.editor import VideoFileClip, AudioFileClip
from IPython.display import Video as IPythonVideo

# Use the first matched pair for demonstration (adjust if necessary)
if video_files and audio_files:
    # Here, we use the first matched pair from the Hungarian algorithm
    idx_video = row_ind[0]
    idx_audio = col_ind[0]
    
    video_path = video_files[idx_video]
    audio_path = audio_files[idx_audio]
    
    print("Combining Video:", os.path.basename(video_path), "with Audio:", os.path.basename(audio_path))
    
    # Load video and audio clips
    video_clip = VideoFileClip(video_path)
    audio_clip = AudioFileClip(audio_path)
    
    # Combine by setting the audio of the video clip
    combined_clip = video_clip.set_audio(audio_clip)
    
    # Save the combined clip to a temporary file (this might take a few seconds)
    combined_output = "combined_sample.mp4"
    combined_clip.write_videofile(combined_output, codec="libx264", audio_codec="aac")
    
    # Display the combined video in the notebook
    display(IPythonVideo(combined_output, embed=True))
else:
    print("No matched video and audio files available for combining.")
```
